```python
from litellm import completion

r = completion(
    model="ollama/gemma3:4b",
    messages=[
        {"role": "system", "content": "you are a helpful assistant. who assistant with scientific meaning? , and guess the scientific meaning words out of arc reactor  "}
    ]
)
print(r.choices[0].message.content)
```

```python
print(f"Input tokens {r.usage.prompt_tokens}")
```

```python
print(f"Output tokens: {r.usage.completion_tokens}")
```

```python
print(f"Total tokens {r.usage.total_tokens}")
```

```python
print(f"Model used: {r.model}")
```

```python
print(f"Total cost :{r._hidden_params["response_cost"]*100:.4f}cents")
```
